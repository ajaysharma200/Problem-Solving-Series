```python
vowels = ['a', 'e', 'i', 'o', 'u']

word = "programming"

count = 0

for character in word:
    if character.lower() in vowels:
        count += 1

print("Number of vowels:", count)
```

## Solution

The program iterates through each character of the word and checks whether it is a vowel.

- A list stores all vowels.
- Each character is compared with the vowel list.
- If a vowel is found, the counter is increased.
- Finally, the total number of vowels is displayed.
