````python
import decimal

string = "12345"

print(decimal.Decimal(string))
print(type(decimal.Decimal(string)))
````

## Answer

A numeric string can also be converted into a `Decimal` object using the `Decimal()` constructor.

In this program:

- The string `"12345"` is passed to `decimal.Decimal()`.
- The resulting object is of type `decimal.Decimal`.

### Output

```text
12345
<class 'decimal.Decimal'>
```
