```python
vowels = ['a', 'e', 'i', 'o', 'u']

word = "programming"

count = 0

for character in word:
    if character.isalpha() and character.lower() not in vowels:
        count += 1

print("Number of consonants:", count)
```

## Solution

The program counts consonants by checking each character in the word.

- It first ensures the character is an alphabet using `isalpha()`.
- It then checks that the character is not a vowel.
- Every valid consonant increases the counter.
